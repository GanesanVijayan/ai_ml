```python
print ("\033[1mAction Performed:\033[0m")
print ("*"*50)
print ("Imported required \033[1mmodules/libraries for Deep Learning Techniques\033[0m")
from tensorflow import keras
from keras.preprocessing.text import Tokenizer
from keras.preprocessing.sequence import pad_sequences
from keras.models import Sequential
from keras.layers import Dense, Embedding, LSTM, Bidirectional
from sklearn.model_selection import train_test_split
import numpy as np
import pandas as pd
```

```python
print ("\033[1mAction Performed:\033[0m")
print ("*"*50)
print ("\t1. \033[1mEMBEDDING INDEX\033[0m created\n")

embeddings_index = {}
f = open('/content/drive/MyDrive/glove.6B.100d.txt')
for line in f:
    values = line.split()
    word = values[0]
    coefs = np.asarray(values[1:], dtype='float32')
    embeddings_index[word] = coefs
f.close()

print('Found \033[1m%s word vectors.\033[0m' % len(embeddings_index))
```

```python
print ("\033[1mAction Performed:\033[0m")
print ("*"*50)
print ("\t1. Displayed Embdeddings Login")
embeddings_index['login']
```

```python
print ("\033[1mAction Performed:\033[0m")
print ("*"*50)
print ("\t1. Created the \033[1mEmbedding Matrix\033[0m")
embedding_matrix = np.zeros((len(word_index) + 1, max_length))
for word, i in word_index.items():
    embedding_vector = embeddings_index.get(word)
    if embedding_vector is not None:
        embedding_matrix[i] = embedding_vector
```

```python
print ("\033[1mAction Performed:\033[0m")
print ("*"*50)
print ("\t1. Created the \033[1mEmbedding Layer\033[0m")
print ("\t2. Built the \033[1mBidirectional LSTM Network\033[0m")
embedding_layer = Embedding(len(word_index) + 1,
                            max_length,
                            weights=[embedding_matrix],
                            input_length=max_length,
                            trainable=False)
embedding_dim = 16
input_length = 100
model = Sequential([
    embedding_layer,
  Bidirectional(LSTM(embedding_dim, return_sequences=True)),
  Bidirectional(LSTM(embedding_dim,)),
  Dense(6, activation='relu'),
  Dense(1, activation='sigmoid')
])
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
```

# **Word2Vec Approach**

```python
print ("\033[1mAction Performed:\033[0m")
print ("*"*50)
print ("\t1. Imported required \033[1mmodules/libraries for Word2Vec\033[0m")
print ("\t1. Performed \033[1mTokenization\033[0m")
import os
import nltk
nltk.download('punkt')
import gensim

dataset['tokenized_corpus'] = dataset.apply(lambda row: nltk.word_tokenize(row['corpus']), axis=1)
```

```python
print ("\033[1mAction Performed:\033[0m")
print ("*"*50)
print ("\t1. Built the \033[1mTokenized corpus\033[0m")
print ("\t2. \033[1mCustom Embeddings Created\033[0m")
print ("\t3. Printed the \033[1mCosine Similarity for a sample\033[0m\n")
word2vec_ip = dataset.tokenized_corpus.tolist()
custom_embeddings = gensim.models.Word2Vec(word2vec_ip, min_count=1,size=300,workers=4)
print("\033[1mSimilarity: \033[0m",custom_embeddings.similarity('issue', 'trouble'))
```

```python
print ("\033[1mAction Performed:\033[0m")
print ("*"*50)
print ("\t1.\033[1m Cleansing the text and embedding corpus with word2vec embedded vectors\033[0m")
docs_vectors = pd.DataFrame() 
stopwords = nltk.corpus.stopwords.words('english') 
for doc in dataset['corpus'].str.lower().str.replace('[^a-z ]', ''): 
    temp = pd.DataFrame()  
    for word in doc.split(' '): 
        if word not in stopwords: 
            try:
                word_vec = custom_embeddings[word] #performing word embedding with word2vec
                temp = temp.append(pd.Series(word_vec), ignore_index = True) 
            except:
                pass
    doc_vector = temp.mean()
    docs_vectors = docs_vectors.append(doc_vector, ignore_index = True)
```

```python
print ("\nShape of the Vectors:")
print ('\033[1mTotal Rows:\033[0m',docs_vectors.shape[0])
print ('\033[1mTotal Columns:\033[0m',docs_vectors.shape[1])
```

```python
print ("\033[1mAction Performed:\033[0m")
print ("*"*50)
print ("\t1.\033[1mDerived the FACTORIZED Labels from primary DataFrame\033[0m")
docs_vectors['Assignment GRP'] = dataset['Assignment GRP']
docs_vectors = docs_vectors.dropna()
```

```python
print ("\033[1mAction Performed:\033[0m")
print ("*"*50)
print ("\t1. Performed \033[1mTRAIN-TEST Split\033[0m")
from sklearn.model_selection import train_test_split

word2vec_X_train, word2vec_X_test, word2vec_y_train, word2vec_y_test = train_test_split(docs_vectors.drop('Assignment GRP', axis = 1), docs_vectors['Assignment GRP'], test_size = 0.2, random_state = 1)
word2vec_X_train.shape, word2vec_y_train.shape, word2vec_X_test.shape, word2vec_y_test.shape

print ("\nShape of the TRAIN-TEST Data:")
print ('\033[1mTRAIN Source Shape:\033[0m',word2vec_X_train.shape)
print ('\033[1mTRAIN Target Shape:\033[0m',word2vec_y_train.shape)
print ('\033[1mTEST Source Shape:\033[0m',word2vec_X_test.shape)
print ('\033[1mTEST Target Shape:\033[0m',word2vec_y_test.shape)
```

```python
from sklearn.ensemble import AdaBoostClassifier
word2vec_ada = AdaBoostClassifier()
word2vec_ada.fit(word2vec_X_train, word2vec_y_train)
test_pred = word2vec_ada.predict(word2vec_X_test)

from sklearn.metrics import accuracy_score
adaboost_score_w2v = accuracy_score(word2vec_y_test, test_pred)
print ("\nAdaBoost Accuracy Score: ",adaboost_score_w2v)

from sklearn.linear_model import LogisticRegression
word2vec_lg = LogisticRegression(solver='liblinear', max_iter=500) 
word2vec_lg.fit(word2vec_X_train, word2vec_y_train)
test_pred = word2vec_lg.predict(word2vec_X_test)

from sklearn.metrics import accuracy_score
lg_score_w2v = accuracy_score(word2vec_y_test, test_pred)
print ("LogisticRegression Accuracy Score: ",lg_score_w2v)
```

```python
print ("\033[1mAction Performed:\033[0m")
print ("*"*50)
print ("\t1. \033[1mWord2Vec Accuracy on AdaBoost and LogisticRegression\033[0m\n")
sns.barplot(["AdaBoost","LogisticRegression"],[adaboost_score_w2v, lg_score_w2v]);
plt.title("Accuracy (Word2Vec) - AdaBoost and LogisticRegression");
```

```python
#Creating pipeline for best performing model - Logistic Regression with Count Vectorizer-NLPAugmented

# connecting to google onedrive
from google.colab import drive
drive.mount('/content/drive')

# importing all required packages
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import re
import nltk
nltk.download('stopwords')
nltk.download('wordnet')
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
from nltk.stem import WordNetLemmatizer
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report
!pip install nlpaug
nltk.download('averaged_perceptron_tagger')
import nlpaug.augmenter.char as nac
import nlpaug.augmenter.word as naw
import nlpaug.augmenter.sentence as nas
import nlpaug.flow as naf
from sklearn.model_selection import KFold
from sklearn.model_selection import cross_val_score
from sklearn.pipeline import Pipeline, make_pipeline
import joblib
import os
import nltk
nltk.download('punkt')
import gensim
import warnings
warnings.filterwarnings("ignore")


# loading data into dataframe
def read_data(file_full_path):
  global dataset
  dataset = pd.read_excel(file_full_path)

def data_preprocessing(dataset, data_usage_for):
  # handling null values
  null_valued_columns = ['Short description','Description']
  for col in null_valued_columns:
      dataset[col]=dataset[col].replace(np.nan,'')

  # dataset filtered for Groups with more than Tickets
  if data_usage_for == 'train':
    dataset = dataset[dataset['Assignment group'].isin(dataset['Assignment group'].value_counts()[dataset['Assignment group'].value_counts()>=1].index)]
    dataset.index = range(0,dataset.shape[0])
    dataset.shape

  # new feature created by concatinating "Short Description" and "Description" columns
  dataset['New Description'] = dataset['Short description']+' '+dataset['Description']
  dataset.head(10)

  # using 'Caller' column in the dataset, created list of caller names as customized STOPWORDS
  list_caller_names = []
  for names in dataset['Caller']:
    list_caller_names.extend(names.split())
  list_caller_names = list(set(list_caller_names))
  # print (list_caller_names[:10])

  # dropping duplicate rows
  dataset = dataset.drop('Caller',axis=1).drop_duplicates()
  dataset.index = range(0,dataset.shape[0])
  dataset.head(10)

  # cleansing overall dataset
  corpus = []
  lemmatizer = WordNetLemmatizer()
  for i in range(0, dataset.shape[0]):
      try:
        description = re.sub('[^a-zA-Z]', ' ', dataset['New Description'][i])
        description = description.lower()
        description = description.split()
        ps = PorterStemmer()
        description = [word for word in description if not word in set(stopwords.words('english'))]
        description = [lemmatizer.lemmatize(word) for word in description if not word in list_caller_names]
        description = ' '.join(description)
        corpus.append(description)
      except:
        print (i,dataset['New Description'][i])
  # creating a new feature called 'corpus' with cleansed text
  # print(len(corpus))
  # print(dataset.shape)
  dataset['corpus']=corpus
  return dataset

def nlp_aug_data_split(dataset):
  # checking assignment groups
  print ("List of Unique Groups: ",dataset['Assignment group'].unique())
  print ("Number of Unique Groups: ",dataset['Assignment group'].nunique())

  # encoding target column 'Assignment GRP' as numerical class
  dataset['Assignment GRP']=pd.factorize(dataset['Assignment group'])[0]
  dataset.head(10)

  # splitting minority and majority classes for nlp augmentation
  dataset_ge_100 = dataset['Assignment GRP'].value_counts()[dataset['Assignment GRP'].value_counts()>=100]
  dataset_lt_100 = dataset['Assignment GRP'].value_counts()[dataset['Assignment GRP'].value_counts()<100]

  #Assignment GRP with greater than or equal to '100' records
  dataset_ge_100.index.tolist()

  #Assignment GRP with lesser than '100' records
  dataset_lt_100.index.tolist()
  global df_ge_100
  global df_lt_100
  df_ge_100 = dataset[dataset['Assignment GRP'].isin(dataset_ge_100.index.tolist())]
  df_lt_100 = dataset[dataset['Assignment GRP'].isin(dataset_lt_100.index.tolist())]
  df_ge_100 = df_ge_100.reset_index()
  df_lt_100 = df_lt_100.reset_index()
  print("Majority class dataset: ", df_ge_100.shape)
  print("Minority class dataset: ", df_lt_100.shape)
  # print("Majority class dataset: ", df_ge_100.columns)
  # print("Minority class dataset: ", df_lt_100.columns)

# text augmentation - synonyms & antonyms generation
def text_augmentation(aug_type='both', aug_sent_cnt=5, corpus_txt='', corpus_txt_assign_grp=''):
    column_names = ['org_aug_text', 'corpus', 'Assignment GRP']
    augmented_dataset = pd.DataFrame(columns = column_names)
    df_temp = pd.DataFrame({"org_aug_text":['orginal'], "corpus":[corpus_txt], "Assignment GRP":[corpus_txt_assign_grp]})
    augmented_dataset = augmented_dataset.append(df_temp, ignore_index = True)
    if aug_type == 'syn':
      aug_syn = naw.SynonymAug(aug_src='wordnet')
      for i in range(aug_sent_cnt):
          augmented_text = aug_syn.augment(corpus_txt)
          df_temp = pd.DataFrame({"org_aug_text":['augmented'], "corpus":[augmented_text], "Assignment GRP":[corpus_txt_assign_grp]})
          augmented_dataset = augmented_dataset.append(df_temp, ignore_index = True)
    elif aug_type == 'ant':
      aug = naw.AntonymAug()
      for i in range(aug_sent_cnt):
          augmented_text = aug.augment(corpus_txt)
          df_temp = pd.DataFrame({"org_aug_text":['augmented'], "corpus":[augmented_text], "Assignment GRP":[corpus_txt_assign_grp]})
          augmented_dataset = augmented_dataset.append(df_temp, ignore_index = True)
    elif aug_type == 'both':
      aug_syn = naw.SynonymAug(aug_src='wordnet')
      for i in range(aug_sent_cnt):
          augmented_text = aug_syn.augment(corpus_txt)
          df_temp = pd.DataFrame({"org_aug_text":['augmented'], "corpus":[augmented_text], "Assignment GRP":[corpus_txt_assign_grp]})
          augmented_dataset = augmented_dataset.append(df_temp, ignore_index = True)
      aug = naw.AntonymAug()
      for j in range(aug_sent_cnt):
          augmented_text = aug.augment(corpus_txt)
          df_temp = pd.DataFrame({"org_aug_text":['augmented'], "corpus":[augmented_text], "Assignment GRP":[corpus_txt_assign_grp]})
          augmented_dataset = augmented_dataset.append(df_temp, ignore_index = True)
    else:
      print("Invalid aug_type parameter passed. Valid ones are 'syn', 'ant', 'both' where 'both' is default.")
    return augmented_dataset

def perform_nlpaug():
    # performing bi-semantic text augmentation
    augmented_dataset_final = pd.DataFrame(columns = ['org_aug_text', 'corpus', 'Assignment GRP'])
    for row in range(len(df_lt_100[:])):
        text = df_lt_100.corpus[row]
        assign_grp = df_lt_100['Assignment GRP'][row]
        if(pd.notnull(df_lt_100.corpus[row])):
          augmented_dataset_res = text_augmentation('both', 5, text, assign_grp)
        augmented_dataset_final = augmented_dataset_final.append(augmented_dataset_res, ignore_index = True)
    augmented_dataset_final = augmented_dataset_final.drop_duplicates(subset=['corpus', 'Assignment GRP'])

    # appending non-augmented dataframe 'df_ge_100' with augmented dataframe 'augmented_dataset_final'
    global df_augmented_final
    df_augmented_final = df_ge_100[['corpus', 'Assignment GRP']].append(augmented_dataset_final[['corpus', 'Assignment GRP']], ignore_index=True)
    print("Augmented dataset shape: ", df_augmented_final.shape)
    return df_augmented_final

def data_split_vectorize(df_augmented_final):
  # #vectorization is handled in model pipeline below
  # # performing count vectorization
  # cv = CountVectorizer(max_features = 7000)
  # cv_X = cv.fit_transform(df_augmented_final['corpus']).toarray()
  # cv_y = df_augmented_final['Assignment GRP'].astype('int')

  # # performing tf-idf vectorization
  # tfv = TfidfVectorizer(max_features = 7000,ngram_range = (1,2))
  # tfv_X = tfv.fit_transform(df_augmented_final['corpus']).toarray()
  # tfv_y = df_augmented_final['Assignment GRP'].astype('int')

  # # performing word2vec vectorization
  # dataset['tokenized_corpus'] = dataset.apply(lambda row: nltk.word_tokenize(row['corpus']), axis=1)
  # word2vec_ip = dataset.tokenized_corpus.tolist()
  # custom_embeddings = gensim.models.Word2Vec(word2vec_ip, min_count=1,size=300,workers=4)

  # # Cleansing the text and embedding corpus with word2vec embedded vectors
  # docs_vectors = pd.DataFrame() 
  # stopwords = nltk.corpus.stopwords.words('english') 
  # for doc in dataset['corpus'].str.lower().str.replace('[^a-z ]', ''): 
  #     temp = pd.DataFrame()  
  #     for word in doc.split(' '): 
  #         if word not in stopwords: 
  #             try:
  #                 word_vec = custom_embeddings[word] #performing word embedding with word2vec
  #                 temp = temp.append(pd.Series(word_vec), ignore_index = True) 
  #             except:
  #                 pass
  #     doc_vector = temp.mean()
  #     docs_vectors = docs_vectors.append(doc_vector, ignore_index = True)

  # # Derived the FACTORIZED Labels from primary DataFrame
  # docs_vectors['Assignment GRP'] = dataset['Assignment GRP']
  # docs_vectors = docs_vectors.dropna()
  # print("Word2Vec dataframe shape: ", docs_vectors.shape)

  # splitting countvec, tfidfvec, word2vec datasets into train and test sets
  global rand_st
  rand_st = 21
  global X_train, X_test, y_train, y_test
  X_train, X_test, y_train, y_test = train_test_split(df_augmented_final['corpus'], df_augmented_final['Assignment GRP'].astype('int'), stratify = df_augmented_final['Assignment GRP'], test_size = 0.20, random_state = rand_st)
  # X_test, X_holdout, y_test, y_holdout = train_test_split(X_test, y_test, stratify = y_test, test_size = 0.10, random_state = rand_st)
  # cv_X_train, cv_X_test, cv_y_train, cv_y_test = train_test_split(cv_X, cv_y, stratify = cv_y, test_size = 0.20, random_state = rand_st)
  # tf_X_train, tf_X_test, tf_y_train, tf_y_test = train_test_split(tfv_X, tfv_y, stratify = cv_y, test_size = 0.20, random_state = rand_st)
  # word2vec_X_train, word2vec_X_test, word2vec_y_train, word2vec_y_test = train_test_split(docs_vectors.drop('Assignment GRP', axis = 1), docs_vectors['Assignment GRP'], test_size = 0.20, random_state = rand_st)


def model_train_persist_infer(X_train, X_test, y_train, y_test):
  logreg_cv_aug_pipe = Pipeline(memory=None,
          steps=[('CountVectorizer',
                  CountVectorizer(max_features = 7000)),
                  ('LogisticRegression',
                  LogisticRegression(C=1.3, class_weight=None, dual=False,
                                      fit_intercept=True, intercept_scaling=0.5,
                                      l1_ratio=None, max_iter=2050,
                                      multi_class='auto', n_jobs=-1,
                                      penalty='l2', random_state=rand_st,
                                      solver='newton-cg', tol=0.0001, verbose=0,
                                      warm_start=False))], verbose=False)
  logreg_cv_aug_pipe.fit(X_train, y_train)
  logreg_cv_aug_y_pred = logreg_cv_aug_pipe.predict(X_test)
  logreg_cv_aug_y_pred_orig = list(map(str, logreg_cv_aug_y_pred.tolist()))
  logreg_cv_aug_y_pred_orig = list(map(( lambda i: 'GRP_' + i), logreg_cv_aug_y_pred_orig))

  # logreg_cv_aug_y_pred_holdout = logreg_cv_aug_pipe.predict(X_holdout)
  # logreg_cv_aug_y_pred_holdout = list(map(str, logreg_cv_aug_y_pred_holdout.tolist()))
  # logreg_cv_aug_y_pred_holdout = list(map(( lambda i: 'GRP_' + j), logreg_cv_aug_y_pred_holdout))

  # persisting model artifacts in google drive for inter-operability
  joblib.dump(logreg_cv_aug_pipe, "/content/drive/MyDrive/capstone_ticket_assignment_nlp_project/countvec_logit_regul1.pkl")
  logreg_cv_model = joblib.load("/content/drive/MyDrive/capstone_ticket_assignment_nlp_project/countvec_logit_regul1.pkl")

  # performing cross validation to evaluate the model performance with 10 cycles
  logreg_cv_aug_pipe_crossval = make_pipeline(logreg_cv_model)
  cv_results = cross_val_score(logreg_cv_aug_pipe_crossval, X_train, y_train, cv=10)


  print("Logistic Regression-Count Vectorizer-NLPAugmented - 10-fold cross validation results: ", cv_results)
  print("Logistic Regression-Count Vectorizer-NLPAugmented - 10-fold cross validation mean accuracy: ", cv_results.mean())
  print("Logistic Regression-Count Vectorizer-NLPAugmented - 10-fold cross validation average standard deviation: ", cv_results.std())
  print("Logistic Regression-Count Vectorizer-NLPAugmented - Train accuracy: ", logreg_cv_aug_pipe.score(X_train, y_train))
  print("Logistic Regression-Count Vectorizer-NLPAugmented - Test accuracy: ", accuracy_score(y_test, logreg_cv_aug_y_pred))
  # print("Logistic Regression-Count Vectorizer-NLPAugmented - Holdout accuracy: ", accuracy_score(y_test, logreg_cv_aug_y_pred_holdout))

  print("Logistic Regression-Count Vectorizer-NLPAugmented - Model prediction - Input: ", X_test.tolist())
  print("Logistic Regression-Count Vectorizer-NLPAugmented - Model prediction - Output: ", logreg_cv_aug_y_pred_orig)

  cv_lg_cm = confusion_matrix(y_test, logreg_cv_aug_y_pred)
  fig = plt.figure(figsize=(75,75));
  sns.heatmap(cv_lg_cm, annot=True,annot_kws={'size': 40});
  plt.title("Logistic Regression-Count Vectorizer-NLPAugmented",fontsize=50);
  print ("Classification Report: Logistic Regression-Count Vectorizer-NLPAugmented");
  print (classification_report(y_test, logreg_cv_aug_y_pred, zero_division=0))
```
