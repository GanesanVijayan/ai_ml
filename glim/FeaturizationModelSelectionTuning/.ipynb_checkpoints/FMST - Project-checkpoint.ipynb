```python
print ("\033[1mAction Performed:\033[0m")
print ("*"*50)
print ("1) \033[1m'Lower Whisker','Min','Quartile 1','Quartile 3','IQR','Max','Upper Whisker'\033[0m values calculated\
for each feature.\n2) Presence of \033[1m'Outliers' & 'Outliers Count' \033[0m is also estimated\n")
five_metrics = df.describe().transpose()
metrics_df = pd.DataFrame()
for cols in five_metrics.index:
    iqr_df = pd.DataFrame()
    iqr_df['Feature'] = [cols]
    iqr_df['Quartile 1'] = [five_metrics.loc[cols]['25%']]
    iqr_df['Quartile 3'] = [five_metrics.loc[cols]['75%']]
    iqr_df['IQR'] = (iqr_df['Quartile 3']-iqr_df['Quartile 1'])*1.5
    iqr_df['Lower Whisker'] = iqr_df['Quartile 1']-iqr_df['IQR']
    iqr_df['Upper Whisker'] = iqr_df['Quartile 3']+iqr_df['IQR']
    iqr_df['Min'] = [five_metrics.loc[cols]['min']]
    iqr_df['Max'] = [five_metrics.loc[cols]['max']]
    if (iqr_df['Min'][0] < iqr_df['Lower Whisker'][0]) or (iqr_df['Max'][0] > iqr_df['Upper Whisker'][0]):
        iqr_df['Outliers']=['Present']
        iqr_df['Outliers Count'] = [(len(df[(df[cols]<iqr_df['Lower Whisker'][0]) | (df[cols]>iqr_df['Upper Whisker'][0])]))]
    else:
        iqr_df['Outliers']=['Not Present']
        iqr_df['Outliers Count'] = [0]
    metrics_df = pd.concat([metrics_df,iqr_df])
    
metrics_df.index = range(0,len(metrics_df))
metrics_df[['Feature','Lower Whisker','Min','Quartile 1','Quartile 3','IQR','Max','Upper Whisker','Outliers','Outliers Count']]
```

```python
from sklearn.model_selection import train_test_split

print ("\n\033[1mAction Performed:\033[0m")
print ("\t1) Dropping target column - \033[1m'Strength'\033[0m and stored other features in variable 'X'")
print ("\t2) The target column \033[1mStrength\033[0m stored in variable 'y'")

X = df.drop(['strength'],axis=1)
y = df['strength']

X_train,X_test,y_train,y_test = train_test_split(X,y,train_size=0.7,random_state=101)

print ("\n\033[1mRecords count after split:\033[0m")
print ("Train Records:",len(X_train))
print ("Test Records:",len(X_test))
```

```python
from sklearn.preprocessing import StandardScaler

#Using Standard Scalar for standardizing data
sc = StandardScaler()
sc_train = pd.DataFrame(sc.fit_transform(X_train))
sc_test = pd.DataFrame(sc.transform(X_test))
sc_train.columns = X_train.columns
sc_test.columns = X_test.columns

print ("\033[1mAction Performed:\033[0m")
print ("*"*50)
print ("1) \033[1mStandard Scalar\033[0m applied on X_train and X_test. \n\
2) Since all of the independent variables are \033[1malmost normally distributed, used only StandardScalar for Scaling\033[0m\
 data\n3) This standardardization is applied on all columns as all are \033[1minteger and continuous \
data with different units\033[0m")
```

```python
from sklearn.decomposition import PCA
pca_all_features = PCA()
pca_all_features.fit(sc_train)
eigen_vectors = pca_all_features.components_
eigen_values = pca_all_features.explained_variance_
per_variance_explained = pca_all_features.explained_variance_ratio_
print ("\033[1mAction Performed:\033[0m")
print ("*"*50)
print ("1) \033[1mPCA\033[0m applied on Scaled Train Data. \n\
2) \033[1mEigen Values and Eigen Vectors\033[0m generated for all features in the dataset\033[0m")
```

```python
print ("\033[1mBar Plot to show the % of variance explained by each Principle Component:\033[0m")
fig = plt.figure(figsize=(15,5))
sns.barplot(list(range(1,9)),per_variance_explained);
plt.xlabel('Principle Components');
plt.ylabel("% Variance Explained");
plt.title("Principle Components Vs % of Variances explained");
```

```python
print ("\n\033[1mPercentage of Variance of each Principle Component:\033[0m\n",per_variance_explained)

print ("\n\033[1mCumulative Percentage of Variance explained by Principle Components\033[0m")
print ('*'*50)
for i in range(1,9):
    print ("Total Variance explained by \033[1mfirst {0} Principle Components\033[0m: {1}".
           format(i,round(sum(per_variance_explained[:i]),2)*100,'%'))

print ("\n\033[1mInference:\033[0m")
print ("*"*50)
print ("1) Since \033[1m7 PCs need to explain 100% variance,\033[0m PCA doesn't hold good for this dataset as the dataset has only EIGHT\
features")
```

```python
print ("\033[1mAction Performed:\033[0m")
print ("*"*50)
print ("1) \033[1mFunction Created\033[0m to calculate all the performance metrics \033[1mMAE, MSE, RMSE, R-SQUARED & \
ADJ.R-SQUARED\033[0m values")
from sklearn.metrics import mean_absolute_error,mean_squared_error
def reg_metrics(model_name,dataset,rows,cols,actual,predicted):
    metrics_dict = {}
    metrics_dict['Model Name']=model_name
    metrics_dict['Dataset']=dataset
    metrics_dict['MAE']=mean_absolute_error(actual,predicted)
    metrics_dict['MSE']=mean_squared_error(actual,predicted)
    metrics_dict['RMSE']=(mean_squared_error(actual,predicted))**(0.5)
    ssr = sum((actual-predicted)**2)       
    sst = sum((actual-np.mean(actual))**2)     
    r_squared = 1 - (float(ssr))/sst
    metrics_dict['R-Squared']=r_squared
    adjusted_r_squared = 1 - (1-r_squared)*(rows-1)/(rows-cols-1)
    metrics_dict['ADJ R-Squared']=adjusted_r_squared
    return (metrics_dict)
```

```python
from sklearn.linear_model import LinearRegression, Ridge, Lasso
model_lm = LinearRegression()
model_lm.fit(sc_train,y_train)
pred_lm = model_lm.predict(sc_test)

ridge_alpha = 0.3
model_ridge = Ridge(alpha=ridge_alpha)
model_ridge.fit(sc_train,y_train)
pred_ridge = model_ridge.predict(sc_test)

lasso_alpha=0.325
model_lasso = Lasso(alpha=lasso_alpha)
model_lasso.fit(sc_train,y_train)
pred_lasso = model_lasso.predict(sc_test)
print ("\033[1mAction Performed:\033[0m")
print ("*"*50)
print ("1) \033[1mLinear Regression, Ridge and Lasso Regression\033[0m applied on the dataset with random ALPHA values")
```

```python
print ("\033[1mAction Performed:\033[0m")
print ("*"*50)
print ("1) \033[1mPerformance metrics of Linear Regression, Lasso and Ridge\033[0m calculated and \
DataFrame displayed with alpha values\n")

print ("\033[1mInference:\033[0m")
print ("*"*50)
print ("1) All three algorithm almost display the \033[1msame performance\033[0m and it indicates \033[1mslight overfit\033[0m")
pd.set_option('display.max_colwidth',None)
metrics_df = pd.DataFrame()
metrics_df=pd.concat([metrics_df,
           pd.DataFrame(reg_metrics('Linear Regression','Train Dataset',sc_train.shape[0],
                                    sc_train.shape[1],y_train,model_lm.predict(sc_train)),index=[0])])
metrics_df=pd.concat([metrics_df,
           pd.DataFrame(reg_metrics('Linear Regression','Test Dataset',sc_test.shape[0],
                                    sc_test.shape[1],y_test,pred_lm),index=[max(metrics_df.index)+1])])
metrics_df=pd.concat([metrics_df,
           pd.DataFrame(reg_metrics('Ridge Regression (alpha={0})'.format(ridge_alpha),'Train Dataset',sc_train.shape[0],
                                    sc_train.shape[1],y_train,model_ridge.predict(sc_train)),index=[max(metrics_df.index)+1])])
metrics_df=pd.concat([metrics_df,
           pd.DataFrame(reg_metrics('Ridge Regression (alpha={0})'.format(ridge_alpha),'Test Dataset',sc_test.shape[0],
                                    sc_test.shape[1],y_test,pred_ridge),index=[max(metrics_df.index)+1])])
metrics_df=pd.concat([metrics_df,
           pd.DataFrame(reg_metrics('Lasso Regression (alpha={0})'.format(lasso_alpha),'Train Dataset',sc_train.shape[0],
                                    sc_train.shape[1],y_train,model_lasso.predict(sc_train)),index=[max(metrics_df.index)+1])])
metrics_df=pd.concat([metrics_df,
           pd.DataFrame(reg_metrics('Lasso Regression (alpha={0})'.format(lasso_alpha),'Test Dataset',sc_test.shape[0],
                                    sc_test.shape[1],y_test,pred_lasso),index=[max(metrics_df.index)+1])])
metrics_df
```

```python
print ("\033[1mInference:\033[0m")
print ("*"*50)
print ("1) The heatmap shows the \033[1mcoeffients of the fit\033[0m for Linear, Lasso and Ridge\n2) \
Lasso shows that feature \033[1m5 & 6 (Coarseagg and Fineagg) doesnt have any impact in prediction\033[0m")
fig = plt.figure(figsize=(10,2))
sns.heatmap(np.array([model_lm.coef_,model_ridge.coef_,model_lasso.coef_]),annot=True, cmap='coolwarm');
```
