```python
#EDA - Checking the presence of outliers in ‘bmi’, ‘age’ and ‘charges columns

#To get IQR for given column
def iqr(col_name):
    oneandhalf_iqr = (df_summary_fin.loc[col_name]['IQR'])*1.5
    outlier_lb = (df_summary_fin.loc[col_name]['25%'])-oneandhalf_iqr
    outlier_ub = (df_summary_fin.loc[col_name]['75%'])+oneandhalf_iqr
    return oneandhalf_iqr,outlier_lb,outlier_ub

#to get upper & lower outliers count, and its total count for given column
def outliers_cnt(col_name):
    lower_outliers = len(df_org[df_org[col_name]<iqr(col_name)[1]])
    upper_outliers = len(df_org[df_org[col_name]>iqr(col_name)[2]])
    total_outliers = len(df_org[df_org[col_name]<iqr(col_name)[1]])+len(df_org[df_org[col_name]>iqr(col_name)[2]])
    return lower_outliers, upper_outliers, total_outliers
```

```python
outliers_dict = {"lower_outliers":outliers_cnt('bmi')[0], "upper_outliers":outliers_cnt('bmi')[1],"total_outliers":outliers_cnt('bmi')[2]}
print("Outlier details for 'bmi' is", outliers_dict)
sns.boxplot(df_org['bmi']);
```

```python
outliers_dict = {"lower_outliers":outliers_cnt('age')[0], "upper_outliers":outliers_cnt('age')[1],"total_outliers":outliers_cnt('age')[2]}
print("Outlier details for 'age' is", outliers_dict)
sns.boxplot(df_org['age']);
```

```python
outliers_dict = {"lower_outliers":outliers_cnt('charges')[0], "upper_outliers":outliers_cnt('charges')[1],"total_outliers":outliers_cnt('charges')[2]}
print("Outlier details for 'charges' is", outliers_dict)
sns.boxplot(df_org['charges']);
```

```python
# Do charges of people who smoke differ significantly from the people who don't?

print(f"This is probability distribution analysis of mean - discrete variable 'charges' for independent groups(smokers & non-smokers) - two sample t test assuming two groups don't have same variance")
# Ho: Two groups have the same mean
# Ha: Two groups don't have the same mean (two tail test)

df_smokers = df_org[df_org['smoker']=='yes'].charges
df_non_smokers = df_org[df_org['smoker']=='no'].charges


if df_smokers.std()**2 == df_non_smokers.std()**2:
    print(f"with equal variance")
else:
    print(f"without equal variance")
    
#default level of significance = 5%. Therefore, alpha = 0.05
#level of confidence = 95%

alpha = 0.05
confidence_level = (1-alpha)*100
t_stats, p_val = ttest_ind(df_smokers, df_non_smokers, equal_var=False)
print(t_stats, p_val)
if p_val<alpha:
    print(f"Two groups don't have the same mean at {confidence_level}% confidence level. Ha is true")
else:
    print(f"Two groups have the same mean at {confidence_level}% confidence level. Ho is true")
    
#Note: perform chi-square test to find the 'equal_var' indicator
```

```python
# Does bmi of males differ significantly from that of females

print(f"This is probability distribution analysis of mean - discrete variable 'bmi' for independent groups(male & female) - two sample t test assuming two groups don't have same variance")
# Ho: Two groups have the same mean
# Ha: Two groups don't have the same mean (two tail test)

df_male = df_org[df_org['sex']=='male'].bmi
df_female = df_org[df_org['sex']=='female'].bmi

#perform chi-square test to find the 'equal_var' indicator
# if df_male.std()**2 == df_female.std()**2:
#     print(f"with equal variance")
# else:
#     print(f"without equal variance")
    
alpha = 0.05
confidence_level = (1-alpha)*100
t_stats, p_val = ttest_ind(df_male, df_female, equal_var=False)
print(t_stats, p_val)
if p_val<alpha:
    print(f"Two groups don't have the same mean at {confidence_level}% confidence level. Ha is true")
else:
    print(f"Two groups have the same mean at {confidence_level}% confidence level. Ho is true")
    
#Note: perform chi-square test to find the 'equal_var' indicator
```

```python
# Is the proportion of smokers significantly different in different genders
print(f"This is probability distribution analysis of proportion - categorical variable 'smokers' for groups(male & female) - two sample proportion test")

# Ho: Two groups have the same proportion
# Ha: Two groups don't have the same proportion (two tail test)

df_male = pd.DataFrame(data=df_org[df_org['sex']=='male'].smoker, columns=['smoker'])
df_female = pd.DataFrame(data=df_org[df_org['sex']=='female'].smoker, columns=['smoker'])
df_male_smoker = pd.DataFrame(data=df_male[df_male['smoker']=='yes'], columns=['smoker_yes'])
df_female_smoker = pd.DataFrame(data=df_female[df_female['smoker']=='yes'], columns=['smoker_yes'])

if len(df_male_smoker)/len(df_male) == len(df_female_smoker)/len(df_female):
    print(f"with equal proportion")
else:
    print(f"without equal proportion")
    
alpha = 0.05
confidence_level = (1-alpha)*100

z_stats, p_val = proportions_ztest([len(df_male_smoker),len(df_female_smoker)],[len(df_male),len(df_female)], prop_var=False)
print(z_stats, p_val)

if p_val<alpha:
    print(f"Two groups don't have the same proportion at {confidence_level}% confidence level. Ha is true")
else:
    print(f"Two groups have the same proportion at {confidence_level}% confidence level. Ho is true")
```

```python
#Is the distribution of bmi across women with no children, one child and two children, the same

print(f"This is probability distribution analysis of variance - contineous variable 'bmi' for 3 groups(female-no child, 1 child, 2 children) & female) - 3 sample oneway ANOVA test")

# Ho: Three groups have the same mean
# Ha: Atleast one group's mean is not equal to other group

alpha = 0.05
confidence_level = (1-alpha)*100
df_female = df_org[df_org['sex']=='female']
df_female_0c = df_female[df_female['children']==0]
df_female_1c = df_female[df_female['children']==1]
df_female_2c = df_female[df_female['children']==2]

f_stats, p_val = f_oneway(df_female_0c.bmi.tolist(), df_female_1c.bmi.tolist(), df_female_2c.bmi.tolist())
print(f_stats, p_val)
if p_val<alpha:
    print(f"Atleast one group's variance is not equal to other group at {confidence_level}% confidence level. Ha is true")
else:
    print(f"Three groups have the same variance at {confidence_level}% confidence level. Ho is true")
```
