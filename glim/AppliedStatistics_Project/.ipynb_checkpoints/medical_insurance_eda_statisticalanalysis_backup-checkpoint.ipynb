```python
def skew_type(skew_val, col_name):
    if -0.5 <= skew_val <= 0.5:
        print(f"Skewness value '{skew_val}' is between -0.5 and 0.5. '{col_name}' data seems to be approximately normally distributed as both the sides are equal")
    elif -0.5 > skew_val >-1.0 or 0.5 < skew_val <1.0:
        print(f"Skewness value '{skew_val}' is between -1.0 and -0.5 or between 0.5 to 1.0. '{col_name}' data seems to be moderately right skewed")
    else:
        print(f"Skewness value '{df_org.charges.skew()}' is less than -1.0 or greater than 1.0. '{col_name}' data seems to be extremely right skewed")
```

```python
n = df_org['charges'].count()
```

```python
m2 = ((df_org['charges']-df_org['charges'].mean())**2).sum()/len(alist)
```

```python
m3 = ((df_org['charges']-df_org['charges'].mean())**2).sum()/len(alist)
```

```python
import numpy as np
import numpy.ma as ma
```

```python
vals = ma.where(m2==0, 0, m3/m2**1.5)
```

```python
vals
```

```python
can_correct = (n > 2) & (m2 > 0)
if can_correct.any():
    m2 = np.extract(can_correct, m2)
    m3 = np.extract(can_correct, m3)
    nval = ma.sqrt((n-1.0)*n)/(n-2.0)*m3/m2**1.5
    np.place(vals, can_correct, nval)
print(vals)
```
